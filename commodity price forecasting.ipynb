```python
# ==========================================
# STEP 1: DATE STANDARDIZATION
# ==========================================

# Create working copies
ohlcv_df = silver_daily.copy()
macro_df = silver_macro.copy()
sentiment_df = silver_sentiment.copy()
futures_df = silver_futures.copy()
supply_demand_df = silver_supply_demand.copy()

# Convert date columns to datetime
ohlcv_df["Date"] = pd.to_datetime(
    ohlcv_df["Date"], errors="coerce"
)

macro_df["Date"] = pd.to_datetime(
    macro_df["Date"], errors="coerce"
)

sentiment_df["Week_Ending"] = pd.to_datetime(
    sentiment_df["Week_Ending"], errors="coerce"
)

futures_df["Trade_Date"] = pd.to_datetime(
    futures_df["Trade_Date"], errors="coerce"
)

# Check data types
print("========== DATE TYPES ==========")
print("OHLCV:", ohlcv_df["Date"].dtype)
print("Macro:", macro_df["Date"].dtype)
print("Sentiment:", sentiment_df["Week_Ending"].dtype)
print("Futures:", futures_df["Trade_Date"].dtype)
print("Supply & Demand:", supply_demand_df["Year"].dtype)

# Check invalid dates
print("\n========== INVALID DATES ==========")
print("OHLCV:", ohlcv_df["Date"].isna().sum())
print("Macro:", macro_df["Date"].isna().sum())
print("Sentiment:", sentiment_df["Week_Ending"].isna().sum())
print("Futures:", futures_df["Trade_Date"].isna().sum())
```

```python
ohlcv_sorted = ohlcv_df.sort_values("Date").reset_index(drop=True)
macro_sorted = macro_df.sort_values("Date").reset_index(drop=True)

# Merge using pd.merge_asof to align monthly macro data with daily OHLCV without look-ahead bias
merged_daily_macro = pd.merge_asof(
    ohlcv_sorted,
    macro_sorted,
    on="Date",
    direction="backward"
)

print("Merged Daily & Macro Shape:", merged_daily_macro.shape)
print("\nMissing values after merge:\n", merged_daily_macro.isna().sum())

# Display the first few rows of key merged columns
print("\nSample Merged Data:")
print(merged_daily_macro[['Date', 'Close', 'Silver_Price_USD', 'Gold_Price_USD', 'DXY_Index']].head())
```

```python
# ==========================================
# STEP 3: MERGING WEEKLY SENTIMENT DATA
# ==========================================

# Sort sentiment dataframe by week ending date
sentiment_sorted = sentiment_df.sort_values("Week_Ending").reset_index(drop=True)

# Drop redundant price column from sentiment to avoid duplicate/conflicting columns upon merge
sentiment_features = sentiment_sorted.drop(columns=["Silver_Price_Close"])

# Merge weekly sentiment into our daily_macro dataframe using merge_asof
merged_daily_macro_sentiment = pd.merge_asof(
    merged_daily_macro.sort_values("Date"),
    sentiment_features,
    left_on="Date",
    right_on="Week_Ending",
    direction="backward"
)

print("Merged Daily, Macro & Sentiment Shape:", merged_daily_macro_sentiment.shape)
print("\nMissing values after sentiment merge:\n", merged_daily_macro_sentiment.isna().sum()[merged_daily_macro_sentiment.isna().sum() > 0])

# Preview key columns including sentiment metrics
print("\nSample Merged Data with Sentiment:")
print(merged_daily_macro_sentiment[['Date', 'Close', 'DXY_Index', 'News_Sentiment_Score', 'CFTC_NonCommercial_Net', 'Google_Trends_Index']].tail())
```

```python
# ==========================================
# STEP 4: MERGING FRONT-MONTH FUTURES DATA
# ==========================================

# Filter for front-month contracts (Months_To_Expiry == 1) to capture spot-futures dynamics and basis
front_month_futures = futures_df[futures_df["Months_To_Expiry"] == 1].copy()

# Sort by trade date
front_month_futures_sorted = front_month_futures.sort_values("Trade_Date").reset_index(drop=True)

# Select key futures features to merge (drop redundant columns like Spot_Price if already in OHLCV)
futures_features = front_month_futures_sorted[[
    "Trade_Date", "Contract_Month", "Futures_Price", "Basis",
    "Basis_Pct", "Open_Interest", "Volume", "Implied_Carry_Rate_Annual"
]].rename(columns={
    "Volume": "Futures_Volume",
    "Open_Interest": "Futures_Open_Interest"
})

# Merge front-month futures into our main dataframe using merge_asof
master_df = pd.merge_asof(
    merged_daily_macro_sentiment.sort_values("Date"),
    futures_features,
    left_on="Date",
    right_on="Trade_Date",
    direction="backward"
).drop(columns=["Trade_Date"])

print("Master Dataset Shape (OHLCV + Macro + Sentiment + Futures):", master_df.shape)
print("\nMissing values in master dataset:\n", master_df.isna().sum()[master_df.isna().sum() > 0])

# Preview key columns of the integrated master dataframe
print("\nSample Integrated Master Data:")
print(master_df[['Date', 'Close', 'DXY_Index', 'News_Sentiment_Score', 'Basis_Pct', 'Implied_Carry_Rate_Annual']].tail())
```

```python
# ==========================================
# STEP 5: HANDLING INITIAL MISSING VALUES & IMPUTATION
# ==========================================

print("Total rows in master dataset:", len(master_df))
print("Rows before sentiment data starts (2000-2003 approx):", master_df['News_Sentiment_Score'].isna().sum())

# Option A: Filter master dataset to start from when sentiment data becomes available (2006-01-06 onwards)
# This ensures all rows have complete multi-modal features without forward-filling across a 6-year gap.
master_df_filtered = master_df.dropna(subset=['News_Sentiment_Score']).reset_index(drop=True)

print("\nFiltered Master Dataset Shape (from 2006 onwards):", master_df_filtered.shape)
print("Remaining missing values:", master_df_filtered.isna().sum().sum())

# Option B: Alternatively, if you want to keep the full 2000-2025 history for technical/macro models,
# you can keep `master_df` and impute/fill early sentiment values using a baseline or median,
# or use `master_df_filtered` specifically for models requiring sentiment features.
```

```python
# ==========================================
# STEP 6: FEATURE ENGINEERING & TARGET CREATION
# ==========================================

# Create a working copy for feature engineering
df_features = master_df_filtered.copy()

# Sort chronologically just in case
df_features = df_features.sort_values("Date").reset_index(drop=True)

# 1. Moving Averages & Trend Features
df_features["SMA_20"] = df_features["Close"].rolling(window=20).mean()
df_features["SMA_50"] = df_features["Close"].rolling(window=50).mean()
df_features["EMA_20"] = df_features["Close"].ewm(span=20, adjust=False).mean()

# 2. Volatility & Momentum Features
df_features["Rolling_Std_20"] = df_features["Close"].rolling(window=20).std()
df_features["Price_to_SMA20_Ratio"] = df_features["Close"] / df_features["SMA_20"]

# 3. Target Variable: Next Day's Return or Next Day's Close Price
# Let's create a classification target (1 if next day return > 0, else 0) and regression target (next day close)
df_features["Target_Next_Close"] = df_features["Close"].shift(-1)
df_features["Target_Next_Return_Pct"] = df_features["Returns_Pct"].shift(-1)
df_features["Target_Direction"] = (df_features["Target_Next_Return_Pct"] > 0).astype(int)

# Drop rows with NaN values created by rolling windows or shifting
df_model_ready = df_features.dropna().reset_index(drop=True)

print("Final Model-Ready Dataset Shape:", df_model_ready.shape)
print("\nTarget Distribution (Direction):")
print(df_model_ready["Target_Direction"].value_counts(normalize=True))

# Preview feature engineered dataset
print("\nSample Feature-Engineered Data:")
print(df_model_ready[['Date', 'Close', 'SMA_20', 'Rolling_Std_20', 'Target_Next_Close', 'Target_Direction']].tail())
```

```python
# ==========================================
# STEP 7: TRAIN-TEST SPLIT & BASELINE MODEL (RANDOM FOREST)
# ==========================================

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Define features (X) and target (y)
# Exclude non-feature columns like Date and targets
exclude_cols = ['Date', 'Week_Ending', 'Contract_Month', 'Target_Next_Close', 'Target_Next_Return_Pct', 'Target_Direction']
feature_cols = [col for col in df_model_ready.columns if col not in exclude_cols]

X = df_model_ready[feature_cols]
y = df_model_ready['Target_Direction']

# 2. Chronological Train-Test Split (e.g., 80% train, 20% test to prevent temporal data leakage)
split_index = int(len(df_model_ready) * 0.8)

X_train, X_test = X.iloc[:split_index], X.iloc[split_index:]
y_train, y_test = y.iloc[:split_index], y.iloc[split_index:]

print(f"Training set shape: {X_train.shape}")
print(f"Testing set shape: {X_test.shape}")

# 3. Train Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)

# 4. Evaluate Model
y_pred = rf_model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print(f"\nRandom Forest Test Accuracy: {accuracy:.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# 5. Feature Importance Preview
feature_importances = pd.Series(rf_model.feature_importances_, index=feature_cols).sort_values(ascending=False)
print("\nTop 10 Most Important Features:")
print(feature_importances.head(10))
```

```python
# ==========================================
# STEP 9: TRAINING AN XGBOOST CLASSIFIER
# ==========================================

import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report

# 1. Initialize XGBoost Classifier
xgb_model = xgb.XGBClassifier(
    n_estimators=150,
    learning_rate=0.05,
    max_depth=5,
    random_state=42,
    eval_metric='logloss',
    n_jobs=-1
)

# 2. Train the model
xgb_model.fit(X_train, y_train)

# 3. Predict and evaluate
y_pred_xgb = xgb_model.predict(X_test)
xgb_accuracy = accuracy_score(y_test, y_pred_xgb)

print(f"XGBoost Test Accuracy: {xgb_accuracy:.4f}")
print("\nClassification Report (XGBoost):")
print(classification_report(y_test, y_pred_xgb))
```

```python
# ==========================================
# STEP 11: TIME SERIES CROSS-VALIDATION
# ==========================================

from sklearn.model_selection import TimeSeriesSplit
from sklearn.ensemble import RandomForestClassifier
import numpy as np

# 1. Setup TimeSeriesSplit (e.g., 5 folds)
tscv = TimeSeriesSplit(n_splits=5)

rf_cv_scores = []

# 2. Iterate through cross-validation folds
for fold, (train_index, test_index) in enumerate(tscv.split(X)):
    X_tr, X_te = X.iloc[train_index], X.iloc[test_index]
    y_tr, y_te = y.iloc[train_index], y.iloc[test_index]

    # Train Random Forest on fold training set
    fold_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
    fold_model.fit(X_tr, y_tr)

    # Evaluate on fold test set
    score = fold_model.score(X_te, y_te)
    rf_cv_scores.append(score)
    print(f"Fold {fold + 1} Accuracy: {score:.4f}")

print(f"\nMean Time-Series CV Accuracy: {np.mean(rf_cv_scores):.4f}")
print(f"Standard Deviation across Folds: {np.std(rf_cv_scores):.4f}")
```

```python
# ==========================================
# STEP 12: LSTM DEEP LEARNING MODEL
# ==========================================

import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import accuracy_score, classification_report

# 1. Prepare and Scale Features
scaler = MinMaxScaler()
scaled_features = scaler.fit_transform(X)

# 2. Function to Create Sequential Windows (e.g., lookback of 10 days)
def create_sequences(data, labels, lookback=10):
    X_seq, y_seq = [], []
    for i in range(lookback, len(data)):
        X_seq.append(data[i-lookback:i])
        y_seq.append(labels.iloc[i])
    return np.array(X_seq), np.array(y_seq)

LOOKBACK = 10
X_seq, y_seq = create_sequences(scaled_features, y, lookback=LOOKBACK)

# 3. Train-Test Split for Sequential Data (Chronological)
split_idx = int(len(X_seq) * 0.8)
X_train_seq, X_test_seq = X_seq[:split_idx], X_seq[split_idx:]
y_train_seq, y_test_seq = y_seq[:split_idx], y_seq[split_idx:]

print(f"LSTM Train Sequence Shape: {X_train_seq.shape}")
print(f"LSTM Test Sequence Shape: {X_test_seq.shape}")

# 4. Build LSTM Model Architecture
lstm_model = Sequential([
    LSTM(64, return_sequences=True, input_shape=(X_train_seq.shape[1], X_train_seq.shape[2])),
    Dropout(0.2),
    LSTM(32, return_sequences=False),
    Dropout(0.2),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])

lstm_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# 5. Train Model
history = lstm_model.fit(
    X_train_seq, y_train_seq,
    epochs=15,
    batch_size=32,
    validation_data=(X_test_seq, y_test_seq),
    verbose=1
)

# 6. Evaluate LSTM Model
y_pred_prob = lstm_model.predict(X_test_seq)
y_pred_lstm = (y_pred_prob > 0.5).astype(int)

lstm_accuracy = accuracy_score(y_test_seq, y_pred_lstm)
print(f"\nLSTM Test Accuracy: {lstm_accuracy:.4f}")
print("\nClassification Report (LSTM):")
print(classification_report(y_test_seq, y_pred_lstm))
```

```python
# ==========================================
# STEP 13: MODernized LSTM MODEL & TRAINING CURVES
# ==========================================

from tensorflow.keras.layers import Input
import matplotlib.pyplot as plt

# 1. Rebuild LSTM using modern Keras syntax (using Input layer)
lstm_model_v2 = Sequential([
    Input(shape=(X_train_seq.shape[1], X_train_seq.shape[2])),
    LSTM(64, return_sequences=True),
    Dropout(0.3),
    LSTM(32, return_sequences=False),
    Dropout(0.3),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])

lstm_model_v2.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# 2. Train Model and save history
history = lstm_model_v2.fit(
    X_train_seq, y_train_seq,
    epochs=20,
    batch_size=32,
    validation_data=(X_test_seq, y_test_seq),
    verbose=1
)

# 3. Plot Training & Validation Curves
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Loss Plot
axes[0].plot(history.history['loss'], label='Train Loss', color='blue')
axes[0].plot(history.history['val_loss'], label='Validation Loss', color='orange')
axes[0].set_title('Model Loss per Epoch')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].legend()

# Accuracy Plot
axes[1].plot(history.history['accuracy'], label='Train Accuracy', color='blue')
axes[1].plot(history.history['val_accuracy'], label='Validation Accuracy', color='orange')
axes[1].set_title('Model Accuracy per Epoch')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy')
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# STEP 14: ENSEMBLE MODEL (RF + XGB + LSTM)
# ==========================================

import numpy as np
from sklearn.metrics import accuracy_score, classification_report

# 1. Get predicted probabilities from each trained model on the test set
# Note: For Random Forest and XGBoost, we aligned X_test with df_model_ready.
# Make sure X_test corresponds to the same test indices used for LSTM sequences.
# The LSTM test sequence starts at split_idx corresponding to test set indices.

# Align test predictions
y_prob_rf_test = rf_model.predict_proba(X_test.iloc[LOOKBACK:])[:, 1]
y_prob_xgb_test = xgb_model.predict_proba(X_test.iloc[LOOKBACK:])[:, 1]
y_prob_lstm_test = lstm_model_v2.predict(X_test_seq).flatten()

# 2. Average the probabilities (Soft Voting Ensemble)
ensemble_probs = (y_prob_rf_test + y_prob_xgb_test + y_prob_lstm_test) / 3.0
y_pred_ensemble = (ensemble_probs > 0.5).astype(int)

# 3. Align ground truth labels for the sequence window
y_true_ensemble = y_test_seq

# 4. Evaluate Ensemble Performance
ensemble_accuracy = accuracy_score(y_true_ensemble, y_pred_ensemble)
print(f"Ensemble Model Test Accuracy: {ensemble_accuracy:.4f}\n")
print("Classification Report (Ensemble):")
print(classification_report(y_true_ensemble, y_pred_ensemble))
```

```python
# ==========================================
# STEP 15 (CORRECTED): STRATEGY BACKTEST & CUMULATIVE RETURNS
# ==========================================

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# 1. Create a backtest dataframe using the perfectly aligned test index
backtest_df = pd.DataFrame(index=X_test_aligned.index)

backtest_df['Actual_Return'] = y_test_aligned.values if hasattr(y_test_aligned, 'values') else y_test_aligned
backtest_df['Ensemble_Pred'] = y_pred_ensemble

# 2. Safely fetch actual percentage returns using the exact aligned test index
backtest_df['Actual_Pct_Return'] = df_model_ready.loc[X_test_aligned.index, 'Target_Next_Return_Pct'].values

# 3. Strategy Return: If prediction is 1 (Up), take the return; if 0 (Down/Cash), return is 0
backtest_df['Strategy_Return'] = backtest_df['Ensemble_Pred'] * backtest_df['Actual_Pct_Return']

# 4. Calculate Cumulative Returns
backtest_df['Cumulative_Buy_Hold'] = (1 + backtest_df['Actual_Pct_Return']).cumprod()
backtest_df['Cumulative_Strategy'] = (1 + backtest_df['Strategy_Return']).cumprod()

# 5. Plot Cumulative Performance
plt.figure(figsize=(12, 6))
plt.plot(backtest_df.index, backtest_df['Cumulative_Buy_Hold'], label='Buy & Hold Benchmark', color='gray', linestyle='--')
plt.plot(backtest_df.index, backtest_df['Cumulative_Strategy'], label='Ensemble ML Strategy', color='green', linewidth=2)

plt.title('Backtest Performance: Ensemble ML Strategy vs. Buy & Hold')
plt.xlabel('Date / Test Set Index')
plt.ylabel('Cumulative Growth Factor ($1 Invested)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# 6. Print Summary Statistics
total_strategy_return = backtest_df['Cumulative_Strategy'].iloc[-1] - 1
total_bh_return = backtest_df['Cumulative_Buy_Hold'].iloc[-1] - 1

print(f"Total Buy & Hold Return: {total_bh_return * 100:.2f}%")
print(f"Total Ensemble Strategy Return: {total_strategy_return * 100:.2f}%")
```
